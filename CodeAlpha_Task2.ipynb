**5. Data Cleaning**

**Handling Missing Values**



*  The dataset contained non-numeric characters such as currency symbols (₹), commas (,), and percentage signs (%), which were successfully removed and converted into numeric values, making the data suitable for analysis and visualization.
*  Using errors='coerce' helped identify invalid or inconsistent entries by converting them into NaN values, enabling easier detection and handling of missing or incorrect data.



```python
# Function to safely clean and convert a column to numeric
def clean_and_convert(df_col, chars_to_remove):
    # Only apply string operations if the column is currently an object (string) type
    if pd.api.types.is_object_dtype(df_col):
        for char in chars_to_remove:
            df_col = df_col.str.replace(char, '', regex=False)
    # Always attempt to convert to numeric, coercing errors to NaN
    return pd.to_numeric(df_col, errors='coerce')

```



*    The price-related columns (actual_price and discounted_price) were successfully converted from text format containing currency symbols and commas into numeric values, enabling accurate price comparisons and statistical analysis.
*   Customer-related metrics such as rating, rating_count, and discount_percentage were standardized into numeric format, ensuring consistency across the dataset and making it suitable for visualization and correlation analysis.



```python
# Apply cleaning and conversion to relevant columns
df['actual_price'] = clean_and_convert(df['actual_price'], ['₹', ','])
df['discounted_price'] = clean_and_convert(df['discounted_price'], ['₹', ','])
df['discount_percentage'] = clean_and_convert(df['discount_percentage'], ['%'])
df['rating_count'] = clean_and_convert(df['rating_count'], [','])
df['rating'] = clean_and_convert(df['rating'], []) # 'rating' typically doesn't need str.replace but might be object

df.head()
```

**VISUALIZATION 1:**

**Top 10 Most Reviewed products**

```python
top_products = df.nlargest(10, 'rating_count')

plt.figure(figsize=(12,6))
sns.barplot(
    x='rating_count',
    y='product_name',
    data=top_products
)

plt.title('Top 10 Most Reviewed Products')
plt.xlabel('Rating Count')
plt.ylabel('Product')
plt.show()
```



*  These products have the highest customer engagement.
*  Popular products tend to accumulate more reviews and visibility.



**VISUALIZATION 2:**

**Product Rating Distribution**

```python
plt.figure(figsize=(8,5))

sns.histplot(df['rating'], bins=10)

plt.title('Distribution of Product Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')

plt.show()
```



*   Most products have ratings between 4.0 and 4.5.
*   Overall customer satisfaction is high.



**VISUALIZATION 3:**

**Top 10 categories**

```python
top_categories = df['category'].value_counts().head(10)

plt.figure(figsize=(10,6))
sns.barplot(
    x=top_categories.values,
    y=top_categories.index
)

plt.title('Top 10 Product Categories')
plt.xlabel('Number of Products')
plt.ylabel('Category')

plt.show()
```



*   Electronics-related products dominate the dataset.
*   Certain categories contribute heavily to Amazon's product listings.



**VISUALIZATION 4:**

**Disount percentange distribution**

```python
plt.figure(figsize=(8,5))

sns.histplot(df['discount_percentage'], bins=20)

plt.title('Discount Percentage Distribution')
plt.xlabel('Discount Percentage')
plt.ylabel('Frequency')

plt.show()
```



*  Most products offer discounts between 40% and 70%.

**VISUALIZATION 5:**

**Actual Price vs Discounted Price**

```python
top_products = df.nlargest(10, 'actual_price')

plt.figure(figsize=(12,6))

x = range(len(top_products))

plt.bar(x, top_products['actual_price'], label='Actual Price')
plt.bar(x, top_products['discounted_price'], label='Discounted Price')

plt.xticks(x, top_products['product_name'], rotation=90)
plt.title('Actual vs Discounted Price of Top 10 Expensive Products')
plt.ylabel('Price (₹)')
plt.legend()

plt.show()
```



*   Shows the direct price reduction after discounts.
*  Makes discount impact easy to understand.



**VISUALIZATION 6:**

**Savings amount per product**

```python
df['savings'] = df['actual_price'] - df['discounted_price']

top_savings = df.nlargest(10, 'savings')

plt.figure(figsize=(10,6))

sns.barplot(
    x='savings',
    y='product_name',
    data=top_savings
)

plt.title('Top 10 Products with Highest Savings')
plt.xlabel('Savings Amount (₹)')
plt.ylabel('Product')

plt.show()
```



*  Highlights products where customers save the most money.
*  Easier to interpret than a scatter plot.



**VISUALIZATION 7:**

**Correlation Heatmap**

```python
numeric_columns = [
    'actual_price',
    'discounted_price',
    'discount_percentage',
    'rating',
    'rating_count'
]

plt.figure(figsize=(8,6))

sns.heatmap(
    df[numeric_columns].corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')

plt.show()
```



*   Actual and discounted prices show strong correlation.
*   Ratings are not strongly influenced by discounts.
